# Test: π and Tajima's D at fourfold-degenerate sites

All-sites π is higher in the Asian vectors than in Ghanaian *An. gambiae* and *An. funestus* (Figure 2b). One explanation is that the site filters differ between data resources. Fourfold-degenerate sites are in coding regions, which are well assembled and less sensitive to site filtering. So if the Asian vectors are still more diverse at these sites, the difference is more likely real.

Same cohorts, regions and cohort sizes as Figure 2b, computed at all sites and at fourfold-degenerate sites (`site_class="CDS_DEG_4"`). *An. dirus* is excluded because the Adir1.0 site annotations are currently unreadable (zarr v3 format).

```python
from pathlib import Path

import malariagen_data
import pandas as pd

RESULTS_CACHE = Path("~/sea_vector_cache").expanduser()
pd.set_option("display.width", 200)
```

```python
amin1 = malariagen_data.Amin1(results_cache=str(RESULTS_CACHE), show_progress=False)
amin1.add_extra_metadata(pd.read_csv("../supplementary_data/amin1_cohorts.csv"))
as1 = malariagen_data.As1(results_cache=str(RESULTS_CACHE), show_progress=False)
ag3 = malariagen_data.Ag3(results_cache=str(RESULTS_CACHE), show_progress=False)
af1 = malariagen_data.Af1(results_cache=str(RESULTS_CACHE), show_progress=False)

# (label, api, region, cohort query, cohort size, extra kwargs)
COHORTS = [
    *[(f"minimus-{pop}", amin1, "KB663610", f"cam_population == '{pop}'", 28, {}) for pop in ["TD", "PV", "RK1", "RK2"]],
    ("stephensi-Afghanistan/Pakistan", as1, "2RL", "country in ['Afghanistan', 'Pakistan']", 30, {}),
    ("stephensi-Sudan", as1, "2RL", "country == 'Sudan' and location != 'PortSudan2018'", 30, {}),
    ("funestus-Ghana", af1, "2RL", "cohort_admin1_year == 'GH-NP_fune_2017'", 30,
     dict(sample_sets="1229-VO-GH-DADZIE-VMF00095")),
    ("gambiae-Ghana", ag3, "3L", "taxon == 'gambiae'", 30,
     dict(sample_sets="AG1000G-GH", site_mask="gamb_colu_arab")),
]
```

```python
rows = []
for label, api, region, query, cohort_size, kwargs in COHORTS:
    for site_class in [None, "CDS_DEG_4"]:
        stats = api.cohort_diversity_stats(
            cohort=(label, query), cohort_size=cohort_size, region=region,
            site_class=site_class, **kwargs,
        )
        rows.append({
            "cohort": label,
            "sites": "fourfold" if site_class else "all",
            "theta_pi": stats["theta_pi"],
            "theta_pi_ci_low": stats["theta_pi_ci_low"],
            "theta_pi_ci_upp": stats["theta_pi_ci_upp"],
            "tajima_d": stats["tajima_d"],
            "tajima_d_ci_low": stats["tajima_d_ci_low"],
            "tajima_d_ci_upp": stats["tajima_d_ci_upp"],
        })

df = pd.DataFrame(rows)
df.round(4)
```

Side by side, with the ratio of fourfold to all-sites π:

```python
df_wide = df.pivot(index="cohort", columns="sites", values=["theta_pi", "tajima_d"])
df_wide.columns = [f"{stat}_{sites}" for stat, sites in df_wide.columns]
df_wide["pi_ratio_fourfold_to_all"] = df_wide["theta_pi_fourfold"] / df_wide["theta_pi_all"]
df_wide = df_wide.loc[[c[0] for c in COHORTS]]
df_wide.round(4)
```

## Missingness and site-filter checks

Two differences from how the published studies computed π:

1. `cohort_diversity_stats` divides by every site that passes the site filter, including sites with missing genotypes, which contribute reduced or zero diversity. Boddé et al. 2025 (*An. funestus*) first dropped sites with more than 5% missing genotypes in the cohort.
2. Ag3 and Af1 default to decision-tree site filters (`dt_20200416`), while Adir1, Amin1 and As1 use static-cutoff filters. Boddé et al. used the Af1 static-cutoff filter (`sc_20220908`).

Below, π at all sites with the API default, then after dropping sites with >5% missing genotypes, then (for *An. funestus*) with the static-cutoff filter. Also the median of windows of 20,000 sites, which is the summary Boddé et al. plotted.

```python
import allel
import numpy as np


def pi_summaries(api, region, query, cohort_size, kwargs, max_missing=0.05, window_size=20_000):
    kwargs = {"site_mask": "default", **kwargs}
    ac = api.snp_allele_counts(
        region=region, sample_query=query, cohort_size=cohort_size, random_seed=42, **kwargs
    )
    an = ac.sum(axis=1)
    keep = an >= (1 - max_missing) * 2 * cohort_size
    mpd = allel.mean_pairwise_difference(allel.AlleleCountsArray(ac), fill=0)
    mpd_keep = mpd[keep]
    n_windows = len(mpd_keep) // window_size
    windows = mpd_keep[: n_windows * window_size].reshape(n_windows, window_size).mean(axis=1)
    return {
        "pi_all_sites": mpd.mean(),
        "frac_sites_dropped": 1 - keep.mean(),
        "pi_missing_filtered": mpd_keep.mean(),
        "pi_window_median": np.median(windows),
    }


af1_sc = malariagen_data.Af1(
    results_cache=str(RESULTS_CACHE), show_progress=False, site_filters_analysis="sc_20220908"
)

rows = []
for label, api, region, query, cohort_size, kwargs in COHORTS:
    rows.append({"cohort": label, "filter": "default", **pi_summaries(api, region, query, cohort_size, kwargs)})
    if api is af1:
        rows.append({"cohort": label, "filter": "sc_20220908", **pi_summaries(af1_sc, region, query, cohort_size, kwargs)})

df_missing = pd.DataFrame(rows)
df_missing.round(4)
```
