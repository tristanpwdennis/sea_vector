# Figure 2: genetic diversity of *An. dirus* s.l.

(a) Runs of homozygosity (ROH) per sample and (b) nucleotide diversity (π) and Tajima's D per cohort, with comparator populations of *An. minimus*, *An. stephensi*, *An. funestus* and *An. gambiae*. *An. dirus* and *An. minimus* statistics are computed with the Adir1.0 and Amin1.0 data via the `malariagen_data` API.

```python
from concurrent.futures import ThreadPoolExecutor

import matplotlib
import matplotlib.pyplot as plt
import malariagen_data
import pandas as pd
import seaborn as sns
from tqdm.auto import tqdm

from common import (
    init_adir1, init_amin1, RESULTS_CACHE, FIGURES_DIR, REPO_DIR,
    ADIR1_COHORTS, ADIR1_COHORT_LABELS, AMIN1_COHORTS, COUNTRY_COLOURS,
)
```

```python
adir1 = init_adir1(show_progress=False)
amin1 = init_amin1(show_progress=False)
df_samples = adir1.sample_metadata()
```

## Figure 2A: runs of homozygosity

ROH are inferred per sample and per contig with `roh_hmm`, and runs shorter than 100 kb are discarded. fROH is the total length of ROH divided by the total length of the 79 contigs.

This is 540 samples × 79 contigs of HMM fits. Results are cached in `RESULTS_CACHE`, so the first run is slow (hours) and later runs are fast.

```python
ROH_MIN_LENGTH = 100_000
genome_length = sum(len(adir1.genome_sequence(c)) for c in adir1.contigs)


def sample_roh(sample_id):
    return pd.concat([adir1.roh_hmm(sample=sample_id, region=c) for c in adir1.contigs])


with ThreadPoolExecutor(max_workers=8) as pool:
    df_roh = pd.concat(
        tqdm(pool.map(sample_roh, df_samples["sample_id"]), total=len(df_samples))
    )
```

```python
df_roh_long = df_roh.query("roh_length >= @ROH_MIN_LENGTH")

df_roh_summary = (
    df_roh_long.groupby("sample_id")["roh_length"]
    .agg(nROH="size", lROH="sum")
    .reindex(df_samples["sample_id"], fill_value=0)
    .reset_index()
    .merge(df_samples, on="sample_id")
)
df_roh_summary["fROH"] = df_roh_summary["lROH"] / genome_length
df_roh_summary.groupby("country")[["nROH", "fROH"]].describe().T
```

```python
sns.set_theme(style="ticks", context="talk")
fig, ax = plt.subplots(figsize=(6, 4))
sns.scatterplot(data=df_roh_summary, x="fROH", y="nROH", hue="country", palette=COUNTRY_COLOURS, ax=ax)
sns.despine(offset=10, trim=False)
for spine in ax.spines.values():
    spine.set_linewidth(0.8)
ax.tick_params(width=0.8, length=5, labelsize=14)
ax.legend(title="Country", fontsize=14, title_fontsize=16, loc="lower right", frameon=False)
fig.savefig(FIGURES_DIR / "fig2a_roh.svg", bbox_inches="tight")
```

## Figure 2B: π and Tajima's D

95% confidence intervals are from 200 block-jackknife replicates (`n_jack=200`, the API default).

### *An. dirus* s.l. and *An. minimus*

Both are computed on the longest contig of each assembly (KB672490 for Adir1, KB663610 for Amin1) with `cohort_diversity_stats`. Cohort sizes follow the original analysis: 30 for *An. dirus* s.l., and 28 (the smallest population, RK2) for *An. minimus*.

```python
STATS_COLS = ["cohort", "theta_pi", "theta_pi_ci_low", "theta_pi_ci_upp", "tajima_d", "tajima_d_ci_low", "tajima_d_ci_upp"]

df_dirus_thetas = pd.DataFrame([
    adir1.cohort_diversity_stats(
        cohort=(ADIR1_COHORT_LABELS[name], query), cohort_size=30, region=adir1.contigs[0]
    )
    for name, query in ADIR1_COHORTS.items()
])

df_minimus_thetas = pd.DataFrame([
    amin1.cohort_diversity_stats(
        cohort=(f"minimus-{name}", query), cohort_size=28, region=amin1.contigs[0]
    )
    for name, query in AMIN1_COHORTS.items()
])
```

### Comparator populations

*An. gambiae* and *An. funestus* from Tanzania are computed with the Ag3 and Af1 APIs. *An. stephensi* values (Dennis et al. 2025) are read from `supplementary_data/stephensi_thetas.csv`.

```python
ag3 = malariagen_data.Ag3(results_cache=str(RESULTS_CACHE), show_progress=False)
df_gambiae = ag3.diversity_stats(
    sample_query="release == '3.0' and country == 'Tanzania'",
    cohorts="admin1_year",
    cohort_size=10,
    region="3L:15,000,000-41,000,000",
    site_mask="gamb_colu_arab",
    site_class="CDS_DEG_4",
).query("cohort == 'TZ-05_gamb_2015'").assign(cohort="gambiae-Tanzania")

af1 = malariagen_data.Af1(results_cache=str(RESULTS_CACHE), show_progress=False)
df_funestus = af1.diversity_stats(
    sample_query="country == 'Tanzania' & admin1_name == 'Kagera'",
    sample_sets=[
        "1236-VO-TZ-OKUMU-OKFR-TZ-2008",
        "1236-VO-TZ-OKUMU-VMF00248",
        "1236-VO-TZ-OKUMU-VMF00252",
        "1236-VO-TZ-OKUMU-VMF00261",
        "1236-VO-TZ-OKUMU-VMF00090",
    ],
    cohorts="admin1_year",
    cohort_size=10,
    region="2RL:15,000,000-41,000,000",
    site_class="CDS_DEG_4",
).assign(cohort="funestus-Tanzania")
```

```python
df_stephensi = (
    pd.read_csv(REPO_DIR / "supplementary_data" / "stephensi_thetas.csv", index_col=0)
    .T.reset_index(drop=True)
)
df_stephensi["cohort"] = ["stephensi-Afghanistan/Pakistan", "stephensi-Sudan"]
```

```python
df_thetas = pd.concat(
    [d[STATS_COLS] for d in [df_dirus_thetas, df_minimus_thetas, df_stephensi, df_funestus, df_gambiae]],
    ignore_index=True,
)
df_thetas[STATS_COLS[1:]] = df_thetas[STATS_COLS[1:]].astype(float)
df_thetas.to_csv(FIGURES_DIR / "fig2b_diversity_stats.csv", index=False)
df_thetas
```

```python
COHORT_PLOT_ORDER = [
    "baimaii-Bangladesh",
    "dirus_ss-Cambodia",
    "baimaii-Cambodia/Thailand",
    "minimus-TD",
    "minimus-PV",
    "minimus-RK1",
    "minimus-RK2",
    "stephensi-Afghanistan/Pakistan",
    "stephensi-Sudan",
    "funestus-Tanzania",
    "gambiae-Tanzania",
]

COHORT_COLOURS = dict(zip(COHORT_PLOT_ORDER, [
    COUNTRY_COLOURS["Bangladesh"],
    COUNTRY_COLOURS["Cambodia"],
    COUNTRY_COLOURS["Thailand"],
    *["#fdbf6f"] * 4,
    *["#fb9a99"] * 2,
    "#cab2d6",
    "#999999",
]))


def plot_thetas(df, stat, symbol, show_xticks=True):
    sns.set_style("ticks")
    fig, ax = plt.subplots(figsize=(6, 3))

    df_plot = df.set_index("cohort").loc[COHORT_PLOT_ORDER]
    for i, (cohort, row) in enumerate(df_plot.iterrows()):
        colour = COHORT_COLOURS[cohort]
        ax.errorbar(
            i, row[stat],
            yerr=[[row[stat] - row[f"{stat}_ci_low"]], [row[f"{stat}_ci_upp"] - row[stat]]],
            fmt="o", color=colour, markersize=8, capsize=7, linewidth=2,
        )

    ax.spines[["top", "right"]].set_visible(False)
    ax.spines["left"].set_position(("outward", 10))
    ax.set_xticks(range(len(COHORT_PLOT_ORDER)))
    ax.set_xticklabels(COHORT_PLOT_ORDER if show_xticks else [""] * len(COHORT_PLOT_ORDER),
                       rotation=45, ha="right", fontsize=14)
    ax.yaxis.set_major_locator(matplotlib.ticker.MaxNLocator(nbins=4))
    ax.yaxis.set_tick_params(labelsize=14)
    ax.set_ylabel(symbol, fontsize=16)
    plt.subplots_adjust(bottom=0.25)
    return fig
```

```python
fig = plot_thetas(df_thetas, stat="theta_pi", symbol="π", show_xticks=False)
fig.savefig(FIGURES_DIR / "fig2b_theta_pi.svg", bbox_inches="tight")
```

```python
fig = plot_thetas(df_thetas, stat="tajima_d", symbol="Tajima's D")
fig.savefig(FIGURES_DIR / "fig2b_tajima_d.svg", bbox_inches="tight")
```
