# Figure 3: genome-wide selection scans

(a) Fst between *An. dirus* s.l. cohorts, (b) G123 within *An. dirus* s.l. cohorts, and (c) G123 within *An. minimus* cohorts. Also produces the outlier-window gene lists (Tables S2 and S3). All statistics are computed with the Adir1.0 and Amin1.0 data via the `malariagen_data` API.

Neither release has phased haplotypes, so G123 uses `sites="all"`: windows of a fixed number of genomic sites passing the site filters, rather than segregating sites. Note the window sizes differ between species (see `G123_PARAMS`).

```python
from itertools import combinations
from pathlib import Path

import malariagen_data
import matplotlib.pyplot as plt
import pandas as pd
```

## Setup

```python
# Results cache shared by all notebooks, so expensive analyses (PCA, ROH, GWSS)
# are only computed once
RESULTS_CACHE = Path("~/sea_vector_cache").expanduser()
FIGURES_DIR = Path("figures")
FIGURES_DIR.mkdir(exist_ok=True)


# In malariagen_data 15.10.0, Adir1 `variant_allele` arrays are object dtype, which
# zarr cannot write to the results cache (ValueError: missing object_codec for
# object array). Cast them to bytes before caching.
class Adir1(malariagen_data.Adir1):
    def results_cache_set(self, *, name, params, results):
        results = {k: v.astype("S") if v.dtype == object else v for k, v in results.items()}
        super().results_cache_set(name=name, params=params, results=results)
```

```python
adir1 = Adir1(results_cache=str(RESULTS_CACHE), show_progress=False)

# An. minimus cohorts (PV/RK1/RK2/TD) are the population assignments of
# St Laurent et al. 2022, attached as the `cam_population` metadata column
amin1 = malariagen_data.Amin1(results_cache=str(RESULTS_CACHE), show_progress=False)
amin1.add_extra_metadata(pd.read_csv("../supplementary_data/amin1_cohorts.csv"))

AMIN1_COHORTS = {pop: f"cam_population == '{pop}'" for pop in ["TD", "PV", "RK1", "RK2"]}
```

```python
# An. dirus s.l. cohorts, defined from the PCA in Figure 1B. Four samples from
# Cambodian site RS23 cluster with An. baimaii.
RS23_BAIMAII = ["S1214", "S1312", "S1047", "S1371"]

ADIR1_COHORTS = {
    "pca_groupa": f"country == 'Thailand' or partner_sample_id in {RS23_BAIMAII}",
    "pca_groupb": f"country == 'Cambodia' and partner_sample_id not in {RS23_BAIMAII}",
    "pca_groupc": "country == 'Bangladesh'",
}

ADIR1_COHORT_LABELS = {
    "pca_groupa": "baimaii-Cambodia/Thailand",
    "pca_groupb": "dirus_ss-Cambodia",
    "pca_groupc": "baimaii-Bangladesh",
}
```

```python
def concat_contig_scan(api, scan_fn, value_col, contigs=None, **kwargs):
    """Run a per-contig GWSS function (e.g. api.g123_gwss) over contigs and
    return a single long dataframe with columns contig, pos, <value_col>."""
    if contigs is None:
        contigs = api.contigs
    dfs = []
    for contig in contigs:
        x, y = scan_fn(contig=contig, **kwargs)
        dfs.append(pd.DataFrame({"contig": contig, "pos": x, value_col: y}))
    return pd.concat(dfs, ignore_index=True)


def add_genome_position(df, api):
    """Add a cumulative `genome_position` column, ordering contigs by length
    (longest first), and an alternating `shade` index per contig."""
    contig_lengths = pd.Series(
        {c: len(api.genome_sequence(c)) for c in df["contig"].unique()}
    ).sort_values(ascending=False)
    offsets = contig_lengths.cumsum().shift(fill_value=0)
    df = df.copy()
    df["contig"] = pd.Categorical(df["contig"], categories=contig_lengths.index, ordered=True)
    df = df.sort_values(["contig", "pos"])
    df["genome_position"] = df["pos"] + df["contig"].map(offsets).astype(float)
    df["shade"] = df["contig"].cat.codes % 2
    return df


def plot_genome_scan(ax, df, value_col, colours=("lightblue", "steelblue")):
    """Plot a genome-wide scan, alternating colours between contigs."""
    for contig, d in df.groupby("contig", observed=True):
        ax.plot(d["genome_position"], d[value_col], color=colours[d["shade"].iloc[0]], lw=0.8)
    ax.spines[["top", "right"]].set_visible(False)
    ax.set_xlim(0, df["genome_position"].max())
```

## Parameters

```python
# Fst window size, in segregating SNPs. NB the Methods give both "100" and "50,000 bp"
# for Fst windows - check which was used for the published figure.
FST_WINDOW_SIZE = 100

# G123 parameters, per species. These reproduce the published values exactly: for
# An. minimus, every Table S3 value at the Cyp450 locus (TD, KB663810); for
# An. dirus s.l., every Table S2 value for Bangladesh An. baimaii on KB672490.
#   window size: number of sites passing filters per window
#   max cohort size: An. dirus s.l. cohorts are downsampled to 50 (the API default);
#     An. minimus cohorts use all samples (n = 156, 58, 41, 28)
#   outlier quantile: windows above this G123 quantile are outliers (Tables S2/S3).
#     0.95 reproduces Table S3 (and matches the Methods). Table S2 only contains windows
#     well above the 95th percentile, consistent with its ">0.99" legend - confirm.
G123_PARAMS = {
    "adir1": dict(window_size=2000, max_cohort_size=50, outlier_quantile=0.99),
    "amin1": dict(window_size=1000, max_cohort_size=None, outlier_quantile=0.95),
}
```

## Figure 3A: Fst between *An. dirus* s.l. cohorts

```python
fst_scans = {}
for cohort1, cohort2 in combinations(ADIR1_COHORTS, 2):
    label = f"{ADIR1_COHORT_LABELS[cohort1]} vs {ADIR1_COHORT_LABELS[cohort2]}"
    df = concat_contig_scan(
        adir1, adir1.fst_gwss, "fst",
        window_size=FST_WINDOW_SIZE,
        cohort1_query=ADIR1_COHORTS[cohort1],
        cohort2_query=ADIR1_COHORTS[cohort2],
    )
    fst_scans[label] = add_genome_position(df, adir1)

pd.DataFrame({label: df["fst"].median() for label, df in fst_scans.items()}, index=["median Fst"]).T
```

```python
fig, axs = plt.subplots(len(fst_scans), 1, figsize=(12, 2.5 * len(fst_scans)), sharex=True, sharey=True)
for ax, (label, df) in zip(axs, fst_scans.items()):
    plot_genome_scan(ax, df, "fst")
    ax.set_title(label, fontsize=11)
    ax.set_ylabel("Fst")
axs[-1].set_xlabel("Genomic position (cumulative, contigs ordered by length)")
fig.tight_layout()
fig.savefig(FIGURES_DIR / "fig3a_fst.svg")
```

## Figure 3B: G123 within *An. dirus* s.l. cohorts

```python
adir1_g123 = {}
for cohort, query in ADIR1_COHORTS.items():
    df = concat_contig_scan(
        adir1, adir1.g123_gwss, "g123",
        window_size=G123_PARAMS["adir1"]["window_size"],
        sites="all",
        sample_query=query,
        max_cohort_size=G123_PARAMS["adir1"]["max_cohort_size"],
    )
    adir1_g123[cohort] = add_genome_position(df, adir1)
```

```python
fig, axs = plt.subplots(len(adir1_g123), 1, figsize=(12, 2.5 * len(adir1_g123)), sharex=True, sharey=True)
for ax, (cohort, df) in zip(axs, adir1_g123.items()):
    plot_genome_scan(ax, df, "g123")
    ax.set_title(ADIR1_COHORT_LABELS[cohort], fontsize=11)
    ax.set_ylabel("G123")
axs[-1].set_xlabel("Genomic position (cumulative, contigs ordered by length)")
fig.tight_layout()
fig.savefig(FIGURES_DIR / "fig3b_adir1_g123.svg")
```

## Figure 3C: G123 within *An. minimus* cohorts

```python
amin1_g123 = {}
for cohort, query in AMIN1_COHORTS.items():
    df = concat_contig_scan(
        amin1, amin1.g123_gwss, "g123",
        window_size=G123_PARAMS["amin1"]["window_size"],
        sites="all",
        sample_query=query,
        max_cohort_size=G123_PARAMS["amin1"]["max_cohort_size"],
    )
    amin1_g123[cohort] = add_genome_position(df, amin1)
```

```python
# Resistance-associated loci with outlier G123 in An. minimus (Table S3), shaded in Figure 3C
AMIN1_LOCI = {
    "Cyp3": ["AMIN000126"],
    "Coe": ["AMIN006586"],
    "UGT": ["AMIN000166"],
    "Gste": ["AMIN009731", "AMIN016012", "AMIN016014", "AMIN016015", "AMIN016016", "AMIN016013", "AMIN016011", "AMIN016017"],
}


def locus_genome_span(api, df_scan, gene_ids):
    """Cumulative genome start/stop of a set of genes, on the same axis as df_scan."""
    genes = api.genome_features(attributes=["ID"]).query("ID in @gene_ids")
    contig = genes["contig"].iloc[0]
    d = df_scan.query("contig == @contig")
    offset = (d["genome_position"] - d["pos"]).iloc[0]
    return genes["start"].min() + offset, genes["end"].max() + offset


fig, axs = plt.subplots(len(amin1_g123), 1, figsize=(12, 2.5 * len(amin1_g123)), sharex=True, sharey=True)
for ax, (cohort, df) in zip(axs, amin1_g123.items()):
    plot_genome_scan(ax, df, "g123")
    ax.set_title(cohort, fontsize=11)
    ax.set_ylabel("G123")
    for locus, gene_ids in AMIN1_LOCI.items():
        start, stop = locus_genome_span(amin1, df, gene_ids)
        ax.axvspan(start, stop, color="grey", alpha=0.3, lw=3)
        if ax is axs[0]:
            ax.text((start + stop) / 2, 1.02, locus, transform=ax.get_xaxis_transform(), ha="center", fontsize=10)
axs[-1].set_xlabel("Genomic position (cumulative, contigs ordered by length)")
fig.tight_layout()
fig.savefig(FIGURES_DIR / "fig3c_amin1_g123.svg")
```

## Tables S2 and S3: genes in outlier G123 windows

For each cohort, windows with G123 above the species' `outlier_quantile` (see `G123_PARAMS`) are called as outliers, and are intersected with the gene annotations. G123 windows are reported by their midpoint, so a window is taken to span half the distance to each neighbouring window.

```python
def g123_outlier_genes(api, g123_scans, outlier_quantile):
    genes = api.genome_features(attributes=["ID", "description"]).query("type.str.endswith('_gene')", engine="python")
    dfs = []
    for cohort, df in g123_scans.items():
        threshold = df["g123"].quantile(outlier_quantile)
        for contig, d in df.groupby("contig", observed=True):
            pos = d["pos"].to_numpy()
            mid = (pos[1:] + pos[:-1]) / 2
            d = d.assign(
                window_start=[pos[0]] + list(mid),
                window_stop=list(mid) + [pos[-1]],
            ).query("g123 > @threshold")
            contig_genes = genes.query("contig == @contig")
            for _, w in d.iterrows():
                hits = contig_genes.query("end >= @w.window_start & start <= @w.window_stop")
                dfs.append(hits.assign(cohort=cohort, g123=w.g123))
    return pd.concat(dfs, ignore_index=True)[["cohort", "g123", "contig", "start", "end", "ID", "description"]]


df_s2 = g123_outlier_genes(adir1, adir1_g123, G123_PARAMS["adir1"]["outlier_quantile"])
df_s3 = g123_outlier_genes(amin1, amin1_g123, G123_PARAMS["amin1"]["outlier_quantile"])
df_s2.to_csv(FIGURES_DIR / "table_s2_adir1_g123_outlier_genes.csv", index=False)
df_s3.to_csv(FIGURES_DIR / "table_s3_amin1_g123_outlier_genes.csv", index=False)
df_s2.groupby("cohort").size(), df_s3.groupby("cohort").size()
```
